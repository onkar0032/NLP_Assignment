```python
import re
import emoji
import pandas as pd

import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.tokenize import RegexpTokenizer

from nltk.corpus import stopwords

from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer

from nltk import pos_tag

from transformers import AutoTokenizer
```

```python
nlp = spacy.load("en_core_web_sm")

bert_tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-multilingual-cased"
)

print("Models Loaded Successfully")
```

```python
dataset = [

    "Breaking News!! India wins the match 🇮🇳🏏 #Cricket https://news.com",

    "OMG 😂😂 This fake news is spreading everywhere!!",

    "आज मौसम बहुत अच्छा है 😊",

    "COVID-19 cases are increasing rapidly in Europe.",

    "@john Check this out!! https://abc.com #Breaking",

    "Yeh news bilkul fake hai!! 😡😡"

]

dataset
```

```python
def clean_text(text):

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"@\w+", "", text)

    text = text.replace("#", "")

    text = emoji.replace_emoji(text, replace="")

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
clean_corpus = []

for text in dataset:

    clean_corpus.append(clean_text(text))

print(clean_corpus)
```

```python
print("Whitespace Tokenization\n")

for text in clean_corpus:

    print(text.split())
```

```python
regex = RegexpTokenizer(r"\w+")

print("Rule-Based Tokenization\n")

for text in clean_corpus:

    print(regex.tokenize(text))
```

```python
print("Subword Tokenization\n")

for text in clean_corpus:

    print(bert_tokenizer.tokenize(text))
```

```python
stop_words = set(stopwords.words("english"))

for text in clean_corpus:

    tokens = word_tokenize(text)

    filtered = [

        word

        for word in tokens

        if word.lower() not in stop_words

    ]

    print(filtered)
```

```python
porter = PorterStemmer()

for text in clean_corpus:

    words = regex.tokenize(text)

    stems = [

        porter.stem(word)

        for word in words

    ]

    print(stems)
```

```python
snow = SnowballStemmer("english")

for text in clean_corpus:

    words = regex.tokenize(text)

    stems = [

        snow.stem(word)

        for word in words

    ]

    print(stems)
```

```python
lemmatizer = WordNetLemmatizer()

for text in clean_corpus:

    words = regex.tokenize(text)

    lemmas = [

        lemmatizer.lemmatize(word)

        for word in words

    ]

    print(lemmas)
```

```python
for text in clean_corpus:

    tokens = word_tokenize(text)

    print(pos_tag(tokens))
```

```python
for text in clean_corpus:

    doc = nlp(text)

    print("\nSentence:", text)

    for token in doc:

        print(

            token.text,

            "| Lemma:", token.lemma_,

            "| POS:", token.pos_

        )
```

```python
sample = clean_corpus[0]

print("Original Text")

print(sample)

print()

print("-----------NLTK-----------")

tokens = word_tokenize(sample)

print(tokens)

print(pos_tag(tokens))

print()

print("-----------spaCy-----------")

doc = nlp(sample)

print([token.text for token in doc])

print([token.lemma_ for token in doc])

print(

    [

        (token.text, token.pos_)

        for token in doc

    ]

)
```
